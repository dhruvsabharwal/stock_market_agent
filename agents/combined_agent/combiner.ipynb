```python
import os
import pandas as pd
import asyncio

# 1. Load master list
a1 = pd.read_csv("../../raw_tickers/nasdaqlisted.txt", sep='|')
all_tickers = a1['Symbol'].tolist()

detailed_path = 'detailed_csv/combined_analysis_all_tickers.csv'
summary_path = 'combined_csv/combined_analysis_all_tickers.csv'

# 2. Identify tickers that actually finished successfully (have a longName)
completed_tickers = set()
existing_df = None

if os.path.exists(detailed_path):
    try:
        existing_df = pd.read_csv(detailed_path)
        # We only count it as completed if 'longName' is present and not 'N/A' or empty
        if 'longName' in existing_df.columns and 'ticker' in existing_df.columns:
            valid_mask = existing_df['longName'].notnull() & (existing_df['longName'] != 'N/A')
            completed_tickers = set(existing_df.loc[valid_mask, 'ticker'].astype(str).tolist())
            print(f"Found {len(completed_tickers)} successfully analyzed tickers.")
    except Exception as e:
        print(f"Reading existing file failed, will start fresh: {e}")

# 3. Filter for tickers to run (those missing or those that failed fundamental analysis)
tickers_to_run = [t for t in all_tickers if str(t) not in completed_tickers]
print(f"Remaining tickers to analyze: {len(tickers_to_run)}")

if len(tickers_to_run) > 0:
    # 4. Run analysis with conservative batch size
    new_results = await analyze_multiple_stocks(tickers_to_run, batch_size=30)
    
    # 5. Process new results
    new_df = pd.json_normalize(new_results.values())
    new_df = new_df.round(3)
    
    # 6. Combine and Deduplicate
    # We keep the 'new' version if a ticker appears in both (updating old failed entries)
    if existing_df is not None:
        combined_df = pd.concat([existing_df, new_df], ignore_index=True)
        combined_df = combined_df.drop_duplicates(subset=['ticker'], keep='last').reset_index(drop=True)
    else:
        combined_df = new_df
    
    # 7. Save results
    os.makedirs('detailed_csv', exist_ok=True)
    combined_df.to_csv(detailed_path, index=False)
    
    os.makedirs('combined_csv', exist_ok=True)
    available_cols = [c for c in (fundamental_cols + technical_cols + score_cols) if c in combined_df.columns]
    combined_df[available_cols].to_csv(summary_path, index=False)
    
    print(f"Done. Total successful tickers in file: {len(combined_df[combined_df['longName'].notnull()])}")
else:
    print("All tickers already have successful analysis results.")
```
